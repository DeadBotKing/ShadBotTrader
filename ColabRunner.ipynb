#3. Model Test:

```python
from src.utils.DatasetPreparer.Config import Config as BaseImageModelConfig
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import tensorflow_datasets as tfds
import tensorflow_hub as hub
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
import logging
import os

tf.random.set_seed(42)
np.random.seed(42)
tf.keras.backend.clear_session()

tfds.disable_progress_bar()
logger = tf.get_logger()
logger.setLevel(logging.ERROR)

class_names = ['buy', 'sell', 'no_trade']


base_image_model_config = BaseImageModelConfig()

zip_dir_base = os.path.dirname(base_image_model_config.cfg['path_Images'])

base_dir = os.path.join(os.path.dirname(base_image_model_config.cfg['path_Images']))
train_dir = os.path.join(base_dir, 'train')
validation_dir = os.path.join(base_dir, 'valid')

train_buy_dir = os.path.join(train_dir, 'buy')
train_sell_dir = os.path.join(train_dir, 'sell')
train_no_trade_dir = os.path.join(train_dir, 'no_trade')
validation_buy_dir = os.path.join(validation_dir, 'buy')
validation_sell_dir = os.path.join(validation_dir, 'sell')
validation_no_trade_dir = os.path.join(validation_dir, 'no_trade')

num_buy_tr = len(os.listdir(train_buy_dir))
num_sell_tr = len(os.listdir(train_sell_dir))
num_no_trade_tr = len(os.listdir(train_no_trade_dir))

num_buy_val = len(os.listdir(validation_buy_dir))
num_sell_val = len(os.listdir(validation_sell_dir))
num_no_trade_val = len(os.listdir(validation_no_trade_dir))

total_train = num_buy_tr + num_sell_tr + num_no_trade_tr
total_val = num_buy_val + num_sell_val + num_no_trade_val

print('total training buy images:', num_buy_tr)
print('total training sell images:', num_sell_tr)
print('total training no_trade images:', num_no_trade_tr)

print('total validation buy images:', num_buy_val)
print('total validation sell images:', num_sell_val)
print('total validation no_trade images:', num_no_trade_val)
print("--")
print("Total training images:", total_train)
print("Total validation images:", total_val)


BATCH_SIZE = 20  # Number of training examples to process before updating our models variables
IMG_SHAPE  = 225  # Our training data consists of images with width of 150 pixels and height of 150 pixels


train_image_generator = ImageDataGenerator(
											rescale=1./255
											# rotation_range = 10,
											# width_shift_range = 0.4,
											# height_shift_range = 0.2,
											# zoom_range = 0.2,
											# horizontal_flip = True,
											# fill_mode = 'nearest'
											)
validation_image_generator = ImageDataGenerator(rescale=1./255)  # Generator for our validation data

train_data_gen = train_image_generator.flow_from_directory(
															batch_size=BATCH_SIZE,
															directory=train_dir,
															shuffle=True,
															target_size=(IMG_SHAPE ,IMG_SHAPE), #(150,150)
															classes = class_names,
															# class_mode='categorical'
															class_mode='binary'
															)

val_data_gen = validation_image_generator.flow_from_directory(
																batch_size=BATCH_SIZE,
																directory=validation_dir,
																shuffle=True,
																target_size=(IMG_SHAPE ,IMG_SHAPE), #(150,150)
																classes = class_names,
																# class_mode='categorical'
																class_mode='binary'
                                                              )

sample_training_images, _ = next(train_data_gen)

def plotImages(images_arr):
	fig, axes = plt.subplots(2, 5, figsize=(20,20))
	axes = axes.flatten()
	for img, ax in zip(images_arr, axes):
		ax.imshow(img)
	plt.tight_layout()
	plt.show()

plotImages(sample_training_images[:10])  # Plot images 0-4

# image_model = tf.keras.models.load_model("mehrshad_image_net.h5", custom_objects={'KerasLayer':hub.KerasLayer})
from tensorflow.keras.layers import Input
input_tensor = Input(shape=(IMG_SHAPE, IMG_SHAPE, 3))

# model_conv_keras = tf.keras.applications.ConvNeXtXLarge(
#                                                         model_name = "convnext_xlarge",
#                                                         include_top = False,
#                                                         include_preprocessing = False,
#                                                         weights = None,
#                                                         input_tensor = None,
#                                                         input_shape = (IMG_SHAPE, IMG_SHAPE, 3),
#                                                         pooling = None,
#                                                         # classes = 3,
#                                                         # classifier_activation = "softmax",
#                                                     )
# model_conv_keras.trainable = True

model_mobile = tf.keras.applications.MobileNetV2(
													input_shape = (IMG_SHAPE, IMG_SHAPE, 3),
													alpha = 1,
													include_top = True,
													weights = 'imagenet',
													input_tensor = None,
													pooling = False,
													classes = 1,
													classifier_activation = "softmax"
												)

model_mobile.trainable = False

# model = tf.keras.models.load_model('my_checkpoint_2.h5')
# model.trainable = True
# model.summary()

# model.add(tf.keras.layers.Dense(128, activation = 'swish', name = 'dense18'))
# model.add(tf.keras.layers.Dropout(0.5, name = 'drop18'))
# model.add(tf.keras.layers.Dense(64, activation = 'swish', name = 'dense19'))
# model.add(tf.keras.layers.Dropout(0.78, name = 'drop19'))
# model.add(tf.keras.layers.Dense(3, activation = tf.nn.softmax, name = 'dense20'))

# model.summary()

# sys.exit()

# early_stopping = tf.keras.callbacks.EarlyStopping(patience=50)
# model_checkpoint = tf.keras.callbacks.ModelCheckpoint("my_checkpoint.h5", save_best_only=True)

# EPOCHS = 200

# history = model.fit_generator(
#                             train_data_gen,
#                             steps_per_epoch=int(np.ceil(total_train / float(BATCH_SIZE))),
#                             epochs=EPOCHS,
#                             validation_data=val_data_gen,
#                             validation_steps=int(np.ceil(total_val / float(BATCH_SIZE))),
#                             callbacks=[early_stopping, model_checkpoint]
#                             )



model = tf.keras.models.Sequential([
                                    # model_conv_keras,
									# image_model,
									model_mobile
								    # tf.keras.layers.Conv2D(32, (3,3), padding = 'same', activation='tanh', input_shape=(IMG_SHAPE, IMG_SHAPE, 3)),
								    # tf.keras.layers.Dropout(0.1),
								    # tf.keras.layers.MaxPooling2D(2, 2),
								    # # tf.keras.layers.Dropout(0.4),

								    # tf.keras.layers.Conv2D(16, (3,3), padding = 'same', activation='tanh'),
								    # tf.keras.layers.Dropout(0.1),
								    # tf.keras.layers.MaxPooling2D(2,2),

                                    # tf.keras.layers.Conv2D(16, (3,3), padding = 'same', activation='swish'),
                                    # tf.keras.layers.Dropout(0.1),
								    # tf.keras.layers.MaxPooling2D(2,2),

                                    # tf.keras.layers.Conv2D(16, (3,3), padding = 'same', activation='swish'),
								    # tf.keras.layers.Dropout(0.1),
								    # tf.keras.layers.MaxPooling2D(2,2),

                                    # tf.keras.layers.Conv2D(16, (3,3), padding = 'same', activation='relu'),
								    # tf.keras.layers.MaxPooling2D(2,2),

								    # tf.keras.layers.Conv2D(16, (3,3), padding = 'same', activation='elu'),
								    # tf.keras.layers.Conv2D(64, (3,3), padding = 'same', activation='elu'),
								    # tf.keras.layers.MaxPooling2D(2,2),
								    # tf.keras.layers.Dropout(0.7),

								    # tf.keras.layers.Conv2D(32, (3,3), padding = 'same', activation='tanh'),
								    # tf.keras.layers.Conv2D(32, (3,3), padding = 'same', activation='elu'),
								    # tf.keras.layers.MaxPooling2D(2,2),
								    # tf.keras.layers.Dropout(0.7),
									# tf.keras.layers.Dropout(0.4),

								    # tf.keras.layers.Flatten(),
                                    # tf.keras.layers.Dropout(0.1),
								    # tf.keras.layers.Conv1D(kernel_size = 16, filters = 32 , padding = 'causal', activation = 'relu'),
								    # tf.keras.layers.Dense(128, activation = 'swish'),
                                    # tf.keras.layers.Dropout(0.4),
								    # tf.keras.layers.Dense(64, activation = 'swish'),
									# tf.keras.layers.Dropout(0.2),
                                    # tf.keras.layers.Dense(64, activation = 'swish'),
                                    # tf.keras.layers.Dense(32, activation = 'swish'),
									# tf.keras.layers.Dropout(0.2),
                                    # tf.keras.layers.Dense(16, activation = 'swish'),
									# tf.keras.layers.Dropout(0.2),
								    # tf.keras.layers.Dropout(0.05),

								    # tf.keras.layers.Dense(3, activation = tf.nn.softmax)
								])

# lr_schedule = tf.keras.callbacks.LearningRateScheduler(lambda epoch: 1e-7 * 10**(epoch / 20))
optimizer = tf.keras.optimizers.Adamax(learning_rate = 3e-04)

model.compile(
	  			optimizer = optimizer,
              	loss = tf.keras.losses.SparseCategoricalCrossentropy(),
              	metrics=['accuracy']
			)

model.summary()

early_stopping = tf.keras.callbacks.EarlyStopping(patience=200)
model_checkpoint = tf.keras.callbacks.ModelCheckpoint("same_image_checker.h5", save_best_only=True)

EPOCHS = 1000
# sys.exit()

# history = model.fit_generator(
# 							    train_data_gen,
# 							    steps_per_epoch=int(np.ceil(total_train / float(BATCH_SIZE))),
# 							    epochs=EPOCHS,
# 							    # validation_data=val_data_gen,
# 							    # validation_steps=int(np.ceil(total_val / float(BATCH_SIZE))),
# 							    callbacks=[lr_schedule]
# 							)

history = model.fit_generator(
							    train_data_gen,
							    steps_per_epoch=int(np.ceil(total_train / float(BATCH_SIZE))),
							    epochs=EPOCHS,
							    validation_data=val_data_gen,
							    validation_steps=int(np.ceil(total_val / float(BATCH_SIZE))),
							    callbacks=[early_stopping, model_checkpoint]
							)

# plt.semilogx(history.history["lr"], history.history["loss"])
# plt.axis([1e-7, 1e-3, 0, 30])

acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

loss = history.history['loss']
val_loss = history.history['val_loss']

epochs_range = range(EPOCHS)

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.plot(acc, label='Training Accuracy')
plt.plot(val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(loss, label='Training Loss')
plt.plot(val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.savefig('./foo.png')
plt.show()
```
